```python
#DATA VISUALIZATION 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv('customer_shopping_behavior.xlsx')
```

```python
df.head()
```

```python
#Count plot(category Distribution)
sns.countplot(x='Category',data=df)
plt.title('Customer Distribution by Category')
plt.xlabel('Category')
plt.ylabel('Count')
plt.show()
```

```python
#Explanation:
The graph shows how many customers purchesed from each category
#insight:
Helps identify the most popular product category
```

```python
#Bar Plot(Average Purchase by Category)
sns.barplot(x='Category',y='Purchase Amount (USD)',data=df)
plt.title('Average Purchase Amount (USD) by Category')
plt.show()
```

```python
#explanation
Shows the average spending in each category
#insight:
which category generates more revenue per customer
```

```python
#pie chart(Gender Distribution)
df['Gender'].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.title('Gender Distribution')
plt.ylabel('')
plt.show()
```

```python

df['Gender'].value_counts().plot(kind='pie',autopct='%1.1f%%',
    figsize=(6,6),       # makes the chart larger
    startangle=90,       # rotates the chart for better alignment
    colormap='Pastel1',  # adds soft colors
    legend=True          # shows a legend
)
```

```python
#Explanation:
the pie chart shows the percentage of male and female customers
#insight
help to understand customer demographics
```

```python
#Histogram(Purchase Amount Distribution)
sns.histplot(df['Purchase Amount (USD)'],kde=True)
plt.title('Distribution of Purchase Amount')
plt.xlabel('Purchase Amount (USD)')
plt.show()
```

```python
#Explanation:
The histogram shows the how purchase value are spread
#insights:
Most customers spend in a specific range
```

```python
#Scatter plot(Age VS Purchase)
plt.scatter(df['Age'],df['Purchase Amount (USD)'])
plt.xlabel('Age')
plt.ylabel('Purchase Amount (USD)')
plt.title('Age Vs Purchase Amount (USD')
plt.show()
```

```python
#Explanation:
the scatter plot shows relationshop between age and spending
#insight:
checks whether age affects purchase behavior
```

```python
#Box plot(Outlier Detection)
sns.boxplot(x='Purchase Amount (USD)',data=df)
plt.title('Box Plot of Purchase Amount (USD)')
plt.show()
```

```python
#Explanation:
The box plot shows the min ,median ,max and outlier.
#insight:
identifies unusually high or low spending customers.
```

```python
#heat map
# Convert if needed
df['Purchase Amount (USD)'] = pd.to_numeric(df['Purchase Amount (USD)'], errors='coerce')

# Select numeric columns
df_numeric = df.select_dtypes(include=['int64', 'float64'])

# Heatmap
sns.heatmap(df_numeric.corr(), annot=True)
plt.title('Correlation Matrix')
plt.show()
```

```python
#Gender vs Purchase(Comparison)
sns.barplot(x='Gender',y='Purchase Amount (USD)',data=df)
plt.title('Purchase Amount (USD) by Gender')
plt.show()
```

```python
#Explanation
comparision shows spending between males and females
#insight:
identifies which gender spend more
```

```python
### Insights:
- The majority of customers belong to the Clothing  category.
- The average purchase amount is higher in Clothing and footwear   category.
- Gender distribution shows that male customers dominate.
- Most purchase amounts fall within a specific range.
- There is a weak relationship between age and purchase amount.
- Some outliers exist in purchase data indicating high-value customers.
```

```python
### Conclusion:
The data visualization helps in understanding customer behavior and purchasing patterns.
These insights can help businesses improve decision-making, target customers effectively,
and increase sales performance.
```
